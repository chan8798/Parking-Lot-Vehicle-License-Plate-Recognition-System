```python
from ultralytics import YOLO

# 모델 불러오기 (사전학습 모델)
model = YOLO("yolo26s.pt")

# 8. 학습
model.train(
    data="/content/drive/MyDrive/YOLO_backup/license plate.v1i.yolo26/data.yaml",
    epochs=100,
    imgsz=960,
    batch=8,

    degrees=5,
    translate=0.5,
    scale=0.9,
    shear=10,
    perspective=0.0001,

    mosaic=1,

)
```

```python
#훈련 모델 구글 드라이브에 저장
import os
import shutil
from google.colab import drive

# 구글 드라이브 마운트
drive.mount('/content/drive')

# 드라이브 저장 폴더 지정
backup_dir = '/content/drive/MyDrive/YOLO_backup3'
os.makedirs(backup_dir, exist_ok=True)

# runs 폴더 백업
shutil.copytree('/content/runs', os.path.join(backup_dir, 'runs'), dirs_exist_ok=True)
```
